```python

path_list = [
    "/Odyssey/private/j25lee/glonet/gradient_descent_optimic/outputs/forecast_exp/idiot_fullobs_idx3_but5_20260518_130917",
    "/Odyssey/private/j25lee/glonet/gradient_descent_optimic/outputs/forecast_exp/idiot_fullobs_idx3_but7_20260518_130917",
    "/Odyssey/private/j25lee/glonet/gradient_descent_optimic/outputs/forecast_exp/idiot_fullobs_idx3_but9_20260518_194127",
]

target_file = "rmse_time_development_global.png"

output_path = "/Odyssey/private/j25lee/glonet/gradient_descent_optimic/outputs/forecast_exp/"
output_file = f"{target_file[:-4]}.gif"
```

```python
from PIL import Image

# Collect all images
images = []
for exp_path in path_list:
    img_path = Path(exp_path) / target_file
    if img_path.exists():
        img = Image.open(img_path)
        images.append(img)
        print(f"✓ Loaded: {img_path.name}")
    else:
        print(f"✗ Not found: {img_path}")

# Create and save GIF
if images:
    output_full_path = Path(output_path) / output_file
    output_full_path.parent.mkdir(parents=True, exist_ok=True)
    
    # Save GIF with 2000ms per frame, looping
    images[0].save(
        output_full_path,
        save_all=True,
        append_images=images[1:],
        duration=2000,  # milliseconds per frame
        loop=0  # infinite loop
    )
    print(f"\n✓ GIF saved: {output_full_path}")
    print(f"  Frames: {len(images)}")
else:
    print("No images found!")
```

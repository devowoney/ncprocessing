# Plot joli RMSE

```python
import json
import glob
import os
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python

database_list = [
    "/Odyssey/private/j25lee/glonet/gradient_descent_optimic/outputs/forecast_exp/idx3_but0_perfect_obs_20260506_220634",
    "/Odyssey/private/j25lee/glonet/gradient_descent_optimic/outputs/forecast_exp/idx0_but0_osse_obs_20260508_164728",
    "/Odyssey/private/j25lee/glonet/gradient_descent_optimic/outputs/forecast_exp/idx3_but-3_fullobs_20260510_180947",
    "/Odyssey/private/j25lee/glonet/gradient_descent_optimic/outputs/forecast_exp/osse_uncertainIC_idx0_but3_20260519_110445"
]
output_path = "/Odyssey/private/j25lee/glonet/gradient_descent_optimic/outputs/forecast_exp/"
output_file = "rmse_conclusion.csv"
```

```python
from IPython.display import display

def load_rmse_tables(exp_path, region="global"):
    rmse_path = Path(exp_path) / "rmse_regional_analysis.json"
    with rmse_path.open() as handle:
        data = json.load(handle)

    steps = data["metadata"]["steps"]
    variables = data["metadata"]["channel_names"]
    reference = np.array(data["reference_rmse"][region], dtype=float)
    optimized = np.array(data["optimized_rmse"][region], dtype=float)

    gain_abs = reference - optimized
    with np.errstate(divide="ignore", invalid="ignore"):
        gain_pct = np.where(reference != 0, gain_abs / reference * 100.0, np.nan)

    gain_abs_df = pd.DataFrame(gain_abs, index=steps, columns=variables)
    gain_pct_df = pd.DataFrame(gain_pct, index=steps, columns=variables)
    return gain_abs_df, gain_pct_df

region_name = "global"

tables = []
for exp_path in database_list:
    gain_abs_df, gain_pct_df = load_rmse_tables(exp_path, region=region_name)
    exp_name = Path(exp_path).name
    print(f"\n=== {exp_name} | region={region_name} ===")
    display(gain_abs_df)
    display(gain_pct_df)

    gain_abs_long = gain_abs_df.copy()
    gain_abs_long["metric"] = "gain_abs"
    gain_pct_long = gain_pct_df.copy()
    gain_pct_long["metric"] = "gain_pct"
    merged = pd.concat([gain_abs_long, gain_pct_long])
    merged["experiment"] = exp_name
    merged["region"] = region_name
    merged["step"] = merged.index
    tables.append(merged)

if tables:
    output_df = pd.concat(tables, ignore_index=True)
    output_full_path = Path(output_path) / output_file
    output_full_path.parent.mkdir(parents=True, exist_ok=True)
    output_df.to_csv(output_full_path, index=False)
    print(f"Saved: {output_full_path}")
```
